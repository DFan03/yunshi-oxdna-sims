```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
mc_timesteps = []
mc_energy = []

with open("energy.dat", "r") as f:
    lines = f.readlines()

# skip first line
for line in lines[1:]:
    parts = line.split()
    
    timestep = int(parts[0])
    energy = float(parts[1])
    
    mc_timesteps.append(timestep)
    mc_energy.append(energy)

mc_timesteps = np.array(mc_timesteps)
mc_energy = np.array(mc_energy)

plt.figure(figsize=(8,5))

plt.plot(mc_timesteps, mc_energy, color="blue", label="MC")

plt.xlabel("Timestep")
plt.ylabel("Potential Energy")
plt.title("Relax Energy Graph (MC)")
plt.legend()

plt.show()
```

```python
md_timesteps = []
md_energy = []

with open("energy_MD_relax.dat", "r") as f:
    for line in f:
        parts = line.split()

        # remove '.' from timestep (0.3000 -> 03000 -> 3000)
        timestep = int(parts[0].replace(".", ""))
        energy = float(parts[1])

        md_timesteps.append(timestep)
        md_energy.append(energy)

md_timesteps = np.array(md_timesteps)
md_energy = np.array(md_energy)
plt.figure(figsize=(8,5))

plt.plot(md_timesteps, md_energy, color="red", label="MD")

plt.xlabel("Timestep")
plt.ylabel("Potential Energy")
plt.title("Relax Energy Graph (MD)")
plt.legend()

plt.show()
```
