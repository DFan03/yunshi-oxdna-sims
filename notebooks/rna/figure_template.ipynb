```python


import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib import rcParams
import numpy as np

# Set font to Arial
rcParams['font.family'] = 'sans-serif'
rcParams['font.sans-serif'] = ['Arial']

# Define color scheme
COLORS = {
    'wavy': '#1F77B4',   #t  # brighter blue
    'circular': '#FF5733',   #k  # punchy red-orange
    'relaxed': '#FF8C00'  #control# vivid orange
}


def format_figure_for_illustrator(ax, show_right_labels=False, show_top_labels=False):
    """
    Full axis box + ticks on all sides, but labels only on left/bottom
    """

    # --- 1. FULL BOX (all spines visible) ---
    for side in ['top', 'right', 'left', 'bottom']:
        ax.spines[side].set_visible(True)
        ax.spines[side].set_linewidth(1.5)
        ax.spines[side].set_color('black')

    # --- 2. TICKS on all 4 sides ---
    ax.tick_params(
        which='both',
        direction='out',
        length=4,
        width=2,
        top=True,
        right=True,
        labelsize=10
    )

    # --- 3. LABELS only on left & bottom ---
    ax.tick_params(
        axis='x',
        labelbottom=True,
        labeltop=False  # ← no numbers on top
    )
    ax.tick_params(
        axis='y',
        labelleft=True,
        labelright=False  # ← no numbers on right
    )

    # --- 4. Axis label font ---
    ax.xaxis.label.set_fontsize(10)
    ax.yaxis.label.set_fontsize(10)

    return ax

#####example
def create_line_plot(x_data, y_data_dict, x_label='', y_label='', title='', 
                     output_filename='figure.svg', figsize=(4, 3)):

    
    fig, ax = plt.subplots(figsize=figsize, dpi=300)
    
    # Plot data with specifications
    for exp_name, y_data in y_data_dict.items():
        ax.plot(x_data, y_data, 
                marker='o',
                markersize=2,  # 2 points for data
                linewidth=2,
                label=exp_name,
                color=COLORS[exp_name])
    
    # Set labels
    ax.set_xlabel(x_label, fontsize=10, family='Arial')
    ax.set_ylabel(y_label, fontsize=10, family='Arial')
    ax.set_title(title, fontsize=10, family='Arial', pad=10)
    
    # Format for Illustrator
    format_figure_for_illustrator(ax)
    
    # Create legend with proper formatting
    legend = ax.legend(loc='best', frameon=False, fontsize=10)
    for text in legend.get_texts():
        text.set_family('Arial')
        text.set_fontsize(10)
    
    plt.tight_layout()
    
    # Save as SVG (vector format, best for Illustrator)
    plt.savefig(dir+output_filename, format='svg', dpi=300, bbox_inches='tight')
    print(f"Figure saved as: {output_filename}")
    
    return fig, ax



# ============================================================================
# TEST DATA AND EXAMPLES
# ============================================================================

if __name__ == '__main__':
    

    
    # Example 1: Line plot with test data

    x_test = np.linspace(0, 10, 20)
    y_test = {
        'bKS': 2 + 0.5*x_test + np.random.normal(0, 0.3, len(x_test)),
        'end': 1.5 + 0.7*x_test + np.random.normal(0, 0.3, len(x_test)),
        'control': 2.5 + 0.2*x_test + np.random.normal(0, 0.3, len(x_test))
    }
    
    create_line_plot(
        x_test, y_test,
        x_label='Time (hours)',
        y_label='Expression Level',
        title='Gene Expression Over Time',
        output_filename='test_line_plot.svg'
    )
    

   
```
