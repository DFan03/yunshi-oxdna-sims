```python
# %% [markdown]
# # Setup
# Import libraries and set common plotting style

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8")
```

```python
# %% 
# 2a. Scatterplot of twist over time (twist_bend.csv)
# Expected format: timestep, twist_deg, bend_deg

tb_df = pd.read_csv(
    "twist_bend.csv",
    header=None,
    names=["timestep", "twist_deg", "bend_deg"],
    engine="python"
)

# Force numeric columns
for col in ["timestep", "twist_deg", "bend_deg"]:
    tb_df[col] = pd.to_numeric(tb_df[col], errors="coerce")

# Drop rows with invalid twist/bend
tb_df = tb_df.dropna(subset=["twist_deg", "bend_deg"])

# Statistics for twist
twist_mean = tb_df["twist_deg"].mean()
twist_std = tb_df["twist_deg"].std(ddof=1)

fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(tb_df["timestep"], tb_df["twist_deg"], alpha=0.5, s=10, label="Twist (deg)", color = 'darkorange')
ax.axhline(twist_mean, color="red", linestyle="--", label=f"Mean = {twist_mean:.2f}")
ax.axhline(twist_mean + twist_std, color="gray", linestyle=":", label=f"Mean ± SD (SD = {twist_std:.2f})")
ax.axhline(twist_mean - twist_std, color="gray", linestyle=":")

ax.set_xlabel("Timestep")
ax.set_ylabel("Twist (degrees)")
ax.set_title("Twist over time")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# %% 
# 2c. Histogram of twist distribution

fig, ax = plt.subplots(figsize=(8, 5))

ax.hist(tb_df["twist_deg"], bins=30, alpha=0.7, color="darkorange", edgecolor="black")
ax.axvline(twist_mean, color="red", linestyle="--", label=f"Mean = {twist_mean:.2f}")
ax.axvline(twist_mean + twist_std, color="gray", linestyle=":", label=f"Mean ± SD (SD = {twist_std:.2f})")
ax.axvline(twist_mean - twist_std, color="gray", linestyle=":")

ax.set_xlabel("Twist (degrees)")
ax.set_ylabel("Count")
ax.set_title("Distribution of twist (degrees)")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# %% 
# 2b. Scatterplot of bend over time

# Statistics for bend
bend_mean = tb_df["bend_deg"].mean()
bend_std = tb_df["bend_deg"].std(ddof=1)

fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(tb_df["timestep"], tb_df["bend_deg"], alpha=0.5, s=10, label="Bend (deg)", color = 'blue')
ax.axhline(bend_mean, color="red", linestyle="--", label=f"Mean = {bend_mean:.2f}")
ax.axhline(bend_mean + bend_std, color="gray", linestyle=":", label=f"Mean ± SD (SD = {bend_std:.2f})")
ax.axhline(bend_mean - bend_std, color="gray", linestyle=":")

ax.set_xlabel("Timestep")
ax.set_ylabel("Bend (degrees)")
ax.set_title("Bend over time")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# %% 
# 2d. Histogram of bend distribution

fig, ax = plt.subplots(figsize=(8, 5))

ax.hist(tb_df["bend_deg"], bins=30, alpha=0.7, color="blue", edgecolor="black")
ax.axvline(bend_mean, color="red", linestyle="--", label=f"Mean = {bend_mean:.2f}")
ax.axvline(bend_mean + bend_std, color="gray", linestyle=":", label=f"Mean ± SD (SD = {bend_std:.2f})")
ax.axvline(bend_mean - bend_std, color="gray", linestyle=":")

ax.set_xlabel("Bend (degrees)")
ax.set_ylabel("Count")
ax.set_title("Distribution of bend (degrees)")
ax.legend()
plt.tight_layout()
plt.show()
```
